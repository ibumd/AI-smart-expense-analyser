```python
import sqlite3
from datetime import datetime

DATABASE = "expense_tracker.db"


# ==========================================================
# DATABASE SETUP
# ==========================================================

def connect_db():
    return sqlite3.connect(DATABASE)


def create_tables():
    conn = connect_db()
    cursor = conn.cursor()

    cursor.execute("""
        CREATE TABLE IF NOT EXISTS expenses (
            id INTEGER PRIMARY KEY AUTOINCREMENT,
            title TEXT NOT NULL,
            amount REAL NOT NULL,
            category TEXT NOT NULL,
            date TEXT NOT NULL,
            note TEXT
        )
    """)

    cursor.execute("""
        CREATE TABLE IF NOT EXISTS budget (
            id INTEGER PRIMARY KEY,
            income REAL DEFAULT 0,
            monthly_budget REAL DEFAULT 0
        )
    """)

    cursor.execute("""
        INSERT OR IGNORE INTO budget
        (id, income, monthly_budget)
        VALUES (1, 0, 0)
    """)

    conn.commit()
    conn.close()


# ==========================================================
# ADD EXPENSE
# ==========================================================

def add_expense():

    print("\n========== ADD EXPENSE ==========")

    title = input("Enter expense title: ")

    while True:
        try:
            amount = float(input("Enter amount: ₹"))

            if amount <= 0:
                print("Amount must be greater than 0.")
                continue

            break

        except ValueError:
            print("Please enter a valid amount.")

    categories = [
        "Food",
        "Travel",
        "Shopping",
        "Bills",
        "Education",
        "Entertainment",
        "Health",
        "Rent",
        "Other"
    ]

    print("\nCategories:")

    for i, category in enumerate(categories, 1):
        print(f"{i}. {category}")

    while True:
        try:
            choice = int(input("Choose category: "))

            if 1 <= choice <= len(categories):
                category = categories[choice - 1]
                break

            print("Invalid category.")

        except ValueError:
            print("Enter a number.")

    date = input(
        "Enter date (YYYY-MM-DD) "
        f"[Press Enter for today]: "
    )

    if date == "":
        date = datetime.now().strftime("%Y-%m-%d")

    note = input("Enter note (optional): ")

    conn = connect_db()
    cursor = conn.cursor()

    cursor.execute("""
        INSERT INTO expenses
        (title, amount, category, date, note)
        VALUES (?, ?, ?, ?, ?)
    """, (title, amount, category, date, note))

    conn.commit()
    conn.close()

    print("\n✓ Expense added successfully!")


# ==========================================================
# VIEW EXPENSES
# ==========================================================

def view_expenses():

    conn = connect_db()
    cursor = conn.cursor()

    cursor.execute("""
        SELECT id, title, amount, category, date, note
        FROM expenses
        ORDER BY date DESC, id DESC
    """)

    expenses = cursor.fetchall()

    conn.close()

    print("\n================ EXPENSE RECORDS ================")

    if not expenses:
        print("No expenses found.")
        return

    print(
        f"{'ID':<5}"
        f"{'Title':<20}"
        f"{'Amount':<12}"
        f"{'Category':<18}"
        f"{'Date':<15}"
    )

    print("-" * 70)

    for expense in expenses:

        expense_id, title, amount, category, date, note = expense

        print(
            f"{expense_id:<5}"
            f"{title[:18]:<20}"
            f"₹{amount:<10.2f}"
            f"{category:<18}"
            f"{date:<15}"
        )

        if note:
            print(f"      Note: {note}")


# ==========================================================
# DELETE EXPENSE
# ==========================================================

def delete_expense():

    view_expenses()

    try:
        expense_id = int(
            input("\nEnter expense ID to delete: ")
        )

    except ValueError:
        print("Invalid ID.")
        return

    conn = connect_db()
    cursor = conn.cursor()

    cursor.execute(
        "SELECT * FROM expenses WHERE id = ?",
        (expense_id,)
    )

    expense = cursor.fetchone()

    if expense is None:
        print("Expense not found.")

    else:

        confirm = input(
            "Are you sure you want to delete it? (y/n): "
        )

        if confirm.lower() == "y":

            cursor.execute(
                "DELETE FROM expenses WHERE id = ?",
                (expense_id,)
            )

            conn.commit()

            print("✓ Expense deleted successfully.")

    conn.close()


# ==========================================================
# SET INCOME AND BUDGET
# ==========================================================

def set_budget():

    print("\n========== BUDGET SETTINGS ==========")

    while True:

        try:

            income = float(
                input("Enter monthly income: ₹")
            )

            if income < 0:
                print("Income cannot be negative.")
                continue

            break

        except ValueError:
            print("Enter a valid amount.")

    while True:

        try:

            budget = float(
                input("Enter monthly budget: ₹")
            )

            if budget < 0:
                print("Budget cannot be negative.")
                continue

            break

        except ValueError:
            print("Enter a valid amount.")

    conn = connect_db()
    cursor = conn.cursor()

    cursor.execute("""
        UPDATE budget
        SET income = ?, monthly_budget = ?
        WHERE id = 1
    """, (income, budget))

    conn.commit()
    conn.close()

    print("\n✓ Budget settings updated.")


# ==========================================================
# SHOW DASHBOARD
# ==========================================================

def dashboard():

    conn = connect_db()
    cursor = conn.cursor()

    cursor.execute("""
        SELECT income, monthly_budget
        FROM budget
        WHERE id = 1
    """)

    settings = cursor.fetchone()

    income = settings[0]
    budget = settings[1]

    cursor.execute("""
        SELECT COALESCE(SUM(amount), 0)
        FROM expenses
    """)

    total_expenses = cursor.fetchone()[0]

    conn.close()

    balance = income - total_expenses

    print("\n")
    print("=" * 55)
    print("           SMART EXPENSE DASHBOARD")
    print("=" * 55)

    print(f"Monthly Income       : ₹{income:.2f}")
    print(f"Monthly Budget       : ₹{budget:.2f}")
    print(f"Total Expenses       : ₹{total_expenses:.2f}")
    print(f"Remaining Balance    : ₹{balance:.2f}")

    print("-" * 55)

    if budget > 0:

        if total_expenses > budget:

            exceeded = total_expenses - budget

            print(
                f"⚠ WARNING: Budget exceeded by ₹{exceeded:.2f}"
            )

        else:

            remaining = budget - total_expenses

            print(
                f"✓ Budget remaining: ₹{remaining:.2f}"
            )

    print("=" * 55)


# ==========================================================
# CATEGORY ANALYSIS
# ==========================================================

def category_analysis():

    conn = connect_db()
    cursor = conn.cursor()

    cursor.execute("""
        SELECT category,
               SUM(amount) AS total
        FROM expenses
        GROUP BY category
        ORDER BY total DESC
    """)

    results = cursor.fetchall()

    conn.close()

    print("\n========== CATEGORY ANALYSIS ==========")

    if not results:
        print("No expense data available.")
        return

    grand_total = sum(
        row[1] for row in results
    )

    for category, total in results:

        percentage = (
            total / grand_total * 100
            if grand_total > 0
            else 0
        )

        bar_length = int(percentage / 2)

        bar = "█" * bar_length

        print(
            f"{category:<18} "
            f"₹{total:<10.2f} "
            f"{percentage:>6.2f}% "
            f"{bar}"
        )


# ==========================================================
# MONTHLY ANALYSIS
# ==========================================================

def monthly_analysis():

    conn = connect_db()
    cursor = conn.cursor()

    cursor.execute("""
        SELECT substr(date, 1, 7) AS month,
               SUM(amount)
        FROM expenses
        GROUP BY month
        ORDER BY month
    """)

    results = cursor.fetchall()

    conn.close()

    print("\n========== MONTHLY EXPENSE ANALYSIS ==========")

    if not results:
        print("No monthly data available.")
        return

    for month, total in results:

        bar_length = int(total / 100)

        bar = "█" * bar_length

        print(
            f"{month} : "
            f"₹{total:.2f} "
            f"{bar}"
        )


# ==========================================================
# SEARCH EXPENSE
# ==========================================================

def search_expense():

    keyword = input(
        "\nEnter title/category to search: "
    ).strip()

    conn = connect_db()
    cursor = conn.cursor()

    cursor.execute("""
        SELECT id, title, amount, category, date, note
        FROM expenses
        WHERE title LIKE ?
           OR category LIKE ?
           OR note LIKE ?
        ORDER BY date DESC
    """, (
        f"%{keyword}%",
        f"%{keyword}%",
        f"%{keyword}%"
    ))

    results = cursor.fetchall()

    conn.close()

    print("\n========== SEARCH RESULTS ==========")

    if not results:
        print("No matching expenses found.")
        return

    for expense in results:

        print(
            f"\nID       : {expense[0]}"
            f"\nTitle    : {expense[1]}"
            f"\nAmount   : ₹{expense[2]:.2f}"
            f"\nCategory : {expense[3]}"
            f"\nDate     : {expense[4]}"
            f"\nNote     : {expense[5]}"
        )


# ==========================================================
# TOP SPENDING CATEGORY
# ==========================================================

def highest_spending_category():

    conn = connect_db()
    cursor = conn.cursor()

    cursor.execute("""
        SELECT category, SUM(amount) AS total
        FROM expenses
        GROUP BY category
        ORDER BY total DESC
        LIMIT 1
    """)

    result = cursor.fetchone()

    conn.close()

    if result:

        print(
            f"\n💡 Highest spending category: "
            f"{result[0]}"
        )

        print(
            f"   Amount spent: ₹{result[1]:.2f}"
        )

    else:

        print("\nNo expense data available.")


# ==========================================================
# MAIN MENU
# ==========================================================

def main():

    create_tables()

    while True:

        print("\n")
        print("=" * 55)
        print("   SMART EXPENSE TRACKING AND ANALYSIS SYSTEM")
        print("=" * 55)

        print("1. Dashboard")
        print("2. Add Expense")
        print("3. View Expenses")
        print("4. Delete Expense")
        print("5. Set Income & Budget")
        print("6. Category Analysis")
        print("7. Monthly Analysis")
        print("8. Search Expense")
        print("9. Highest Spending Category")
        print("10. Exit")

        print("-" * 55)

        choice = input(
            "Enter your choice (1-10): "
        )

        if choice == "1":
            dashboard()

        elif choice == "2":
            add_expense()

        elif choice == "3":
            view_expenses()

        elif choice == "4":
            delete_expense()

        elif choice == "5":
            set_budget()

        elif choice == "6":
            category_analysis()

        elif choice == "7":
            monthly_analysis()

        elif choice == "8":
            search_expense()

        elif choice == "9":
            highest_spending_category()

        elif choice == "10":

            print("\nThank you for using Smart Expense Tracker!")
            print("Goodbye!")

            break

        else:

            print(
                "\nInvalid choice. "
                "Please select 1-10."
            )


# ==========================================================
# START PROGRAM
# ==========================================================

if __name__ == "__main__":
    main()
```
